## Functions

Functions in Python group resuable code into a  block that can be executed by calling and make programs modular and readable . There are Many Inbuilt Functions in Python like print(), input() etc:

We can create our own functions . To make a function in python we have to use **def keyword and then name the funciton, after this you have to call the function using name and parenthesis**

**FUNCTION PARAMETRS AND ARGUMENTS**

Parameters : Variable Listed inside function definition
Arguments : Value passed to the function

**Types of Argument :**
    3 types of argument are there
        1. **Positional Argument**
        2. **Default Argument**
        3. **Keyword Argument**
