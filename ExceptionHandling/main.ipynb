# EXCEPTION HANDLING


**ERRORS**

Errors occure due to mistakes in the code that prevent it from running. These can be syntax errors or logical errors.

    SYNTAX ERROR :  print("hello world

    INDENDATION ERROR 

**EXCEPTIONS**

Unexpected Events or errors that occurs during the execution of a program , which disrupts the normal flow of program.

EX : print(10/0) ==> Zero Division Error



**EXCEPTION HANDLING**

    try : wraps the block of code that might cause an exception. It must have one expect statement or finally clause

    except : Handle the exception if it occurs

    else : Run code only if no exception occurs

    finally : Run code no matter , whether there's an exception or not

    raise : Manually throw an exception
