```python
## Multilevel Inheritence

class Factory:
    def __init__(self,material,zips):
        self.material = material
        self.zips = zips

class NikeFactory(Factory):
    def __init__(self, material, zips,color):
        super().__init__(material, zips)
        self.color = color

class ReebokFactory(NikeFactory):
    def __init__(self, material, zips, color,pockets):
        super().__init__(material, zips, color)
        self.pockets = pockets


```

```python
# Mtehod Overriding
class Animal:
    def show(self):
        print("Hello World")

class Human(Animal):
    def show(self):  ## Method Overriding -> If you have classes one parent and one child , parent class and child class have same name methods and if the object is calling the method then method will be called of child class, This is a basic Definition of Method Overriding
        print("Sup!?")

obj = Human()
 

obj.show()
```

### Encapsulation

```python
## Encapsulation : Hiding internal details of class and only showing ehat is needed. Keeps data sade from being changes by mistake
class Factory:
    a = "pune"

    def show(self):
        print("hello o am a pune factory")

obj = Factory()

print(obj.a)

obj.a = "mumbai" #we changed a here

print(obj.a)
```

```python
class Factory:
    a = "pune"

    def show(self):
        print("hello o am a pune factory")

class Mumbai(Factory):
    def show2(self):
        print(super().a)

obj = Mumbai()

obj.show2()
## Here also we are easily able to access the attributes and methods These are called public attributes and methods.
```

```python
class Factory:
    _a = "pune" # protected attribute

    def _show(self):
        print("hello o am a pune factory")

class Mumbai(Factory):
    def show2(self):
        print(super()._a)

obj = Mumbai()

obj.show2()
```

```python
# Private Attributes and Methods

# It can not be accesses from outside the class only from inside the class



class Factory:
    __a = "pune" 

    def __show(self):
        print("hello o am a pune factory")

class Mumbai(Factory):
    def show2(self):
        print(super().__a)

obj = Mumbai()
obj.show2() # you will see error as it cannot be accessed
```

```python
class Factory:
    __a = "mumbai"

    def show(self):
        print(Factory.__a)

obj = Factory()

obj.show()

## We are now able to access private class this way
```

```python
## Examples

class Demo:
    def __init__(self):
        self.name = "Public Member"         #Public
        self._age = 21                      #Protected
        self.__salary = 50000               #Private

    def show(self):
        print("Inside the class")
        print("Public",self.name)
        print("Protected",self._age)
        print("Private",self.__salary)

obj = Demo()

obj.show()
```

### Abstraction

```python
## It does not exist in python but we can achieve  it using library

## Used to simplyfing complex systems by focusing on essential features and hiding unnecessary details
## Used to define common interface for diffrent subclasses

from abc import ABC,abstractmethod

class abstract(ABC):
    @abstractmethod
    def perimeter(self):
        pass
    @abstractmethod
    def area(self):
        pass


class Square(abstract):
    def __init__(self,side):
        self.side = side

    def perimeter(self):
        print("Created Perimeter")
    
    def area(self):
        print("Created area  ")

class Circle(abstract):
    def __init__(self, radius):
        self.radius = radius
    
    def perimeter(self):
        print("Created Perimeter")
    
    def area(self):
        print("Created area  ")

obj = Circle(7)

obj2 = Square(12)

```

```python
## Special methods used in python that starts and ends with double underscore
## They automatically gets called when performing certain actions on an object
## Make your objects behave like built in data types


class Animal:
    def __init__(self,name,age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"hello ,how are you {self.name}?"
    
    def __add__(self,other):
        return f"your sum of ages are {self.age + other.age}"


obj = Animal("lion", 12)
obj2 = Animal("dolphin", 14)

print(obj + obj2)
        
```
