### Modules and Packages in Python

#### Module

A **module** is a single Python file (`.py`) that contains reusable code. Modules help organize programs and allow code to be reused across multiple files.

#### Key Points

* A module is a single Python file (`.py`).
* It can contain:

  * Functions
  * Variables
  * Classes
* Code from one module can be imported and used in another Python file.
* Modules improve code organization and reusability.

### Built-in Python Modules

Python provides many ready-to-use modules, such as:

* `math` – Mathematical operations and functions.
* `random` – Generating random numbers and selections.
* `datetime` – Working with dates and times.

---

#### Package

A **package** is a directory (folder) that contains one or more Python modules. A package can also contain sub-packages, allowing large projects to be organized into a structured hierarchy.

##### Key Points

* A package is a collection of related modules.
* It helps organize large codebases efficiently.
* Packages can contain:

  * Multiple modules
  * Sub-packages
* Modules and packages are typically accessed using the `import` and `from ... import ...` statements.

##### Third-Party Packages

In addition to Python's built-in modules, there are many third-party packages that can be installed and used in projects, including:

* `NumPy` – Numerical computing and array operations.
* `Pandas` – Data analysis and data manipulation.
* `Matplotlib` – Data visualization and plotting.

These packages must be installed separately using package managers such as `pip`.

---

#### Summary

* **Module** = A single Python file containing reusable code.
* **Package** = A folder containing one or more modules (and possibly sub-packages).
* Python includes many built-in modules.
* Third-party packages can be installed to extend Python's capabilities.
* `import` and `from ... import ...` are used to access modules and packages in your programs.
