# Conditional Statements

It allows decision-making by execution diffrent blocks of code based on conditions.

ex : you will receive a number from the user if the number is greater than 10 you will do task A and lower you will do task B

## Types of Conditional Statement

3 types 
1. if : executes if the condition is true
2. if-else : executes if True , another if False
3. if-elif-else : Checks multuple conditions in sequence

## Examples

```python
a = int(input("Enter your age"))
if a > 18:
    print("Legal to vote")
else:
    print("not legal to vote")
```

```python
a = int(input("Enter number 1"))
b = int(input("Enter number 2"))
if a==b:
    print("both are same")
elif a>b:
    print("a is greater")
else:
    print("b is greater")
```
