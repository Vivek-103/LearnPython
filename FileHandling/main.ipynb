# FILE HANDLING



**MEANING**

    File Handling means performing CRUD operations on file

    There are methods to perform these in Python

    We have to use open() function to open a file in python

**MODES TO OPEN A FILE**

    r : Read(default) -> file must exist

    w : Write -> create files or overwrites

    a : Append -> adds to end of file

    x : Create -> Create a new file , fails if it exists

```python
p = open(r'main.ipynb')

print(p.read())
```

```python
r = open("work.txt",'w')

r.write("Hello WORLD SUP !!?")

r.close()

```

```python
r = open("work.txt",'a')

r.write("now i am appending to the file")

r.close()
```
