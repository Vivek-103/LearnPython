# Operators in Python
    Operators are symbols that perform on variables and values . Several types of operators are there : 

    Arthmetic Operators : 
        used to perform mathematical operations like addition , substraction , multiplication , division etc :
        There are 7 types of arithmetic operators
        addition +
        subtraction -
        multiplication *
        division /
        Floor division //
        modulus %
        Exponentiation **
    
    Assignment Operator 
        Used to assign values to variables  EX : =

        Compound Assignment Operators combine arthimetic opeations with assignment

        Using compound assignment operators the reassigning
        works better :
        += Add and assign
        -= Subtract and assign
        *= Multiply and assign
        /= Divide and assign
        //= Floor divide and assign
        %= modulus and assign
        **= Exponentiation and assign
    
    Comparison Operator
        Also called relational operators are used to compare 2 values . They always provide boolean result that is True or False

        comparison operators are as follow
         == Equal to
         != Not Equal to
         > Greater than
         < Less than 
         >= Greater than or equal to
         <= Less than or equal to
        Wotks with numbers but we can use it using string as well ,
        Strings will be compating the ASCII values of string.
    
    Logical Operators
        Used to combine multiple conditions and result a Boolean result (True or False)

        There are 3 types of logical operator 
        and -> Return True if both conditions are True
        or -> Returm True if atleast one condition is True
        not -> Reverse the boolean value
