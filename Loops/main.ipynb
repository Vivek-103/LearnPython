# Loops in Python


allows us to use a block of code multiple times without rewriting it .

## 2 Types of Loops

1. For Loop : When you know the number of iterations is Known
2. While Loop : You dont know number of iterations but you know a condition that determines when to stop.

## Terminiologies

1. RANGE FUNCTION : used to generate a sequence of numbers , commonly used in loops . Syntax : range ( start , stop , steps)
if not mentioned start value will be 0 and default steps will be 1 entering stop value is necessary

2. Loops for Numbers : For using loops with numbers you need the range function . EX : for i in range (1,6): ,print (i)

3. Loops for Strings : You can iterate though string in 2 ways:
    a. Using Index values
    b. Iterating directly over the string.
    EX : a = "nature" , for i in range(len(a)): print(a[i])-> iterated over string using index values
    EX : a = "nature" , for chat in a: print(char) -> directly iterate, but this method will give direct access to the characters instead of index values.


```python
# Print hello words 10 times

for i in range(1 , 10):
    print("hello")
```

## EXAMPLES AND PROBLEMS

```python
a = int(input("Enter the range of number"))
for i in range (0,a):
    print("Hello world")
```

```python
## print natural numbers upto n

n = int(input("Enter the n"))
for i in range(0,n):
    print(i+1)
```

```python
## print natural number from n to 1

n = int(input("Enter the n"))
for i in range(n,0,-1):
    print(i)
```

```python
## Print the table 

n = int(input("Enter the number whose table you want"))
for i in range(1,11):
    print("table of" ,n ,"*", i ,"is" , n * i)
```

```python
## Sum upto n terms

n = int(input("Enter the value of n"))
print ("sum of n terms is" , ((n)*(n+1)//2))
```

```python
## Break Continue

for i in range(1,21):
    if i == 15:
        break
    else:
        print(i)
```

```python
## Factorial

n = int(input("Enter the number you want factorial of:"))
fact = 1
for i in range(1,n+1):
    fact = fact * i
print(f"{fact}")


```

```python
## Print all even and acts seperately

n = int(input("Enter the number :"))
even = 0
odd = 0
for i in range(1,n+1):
    if i%2==0:
        even = even + i
    else:
        odd = odd + i
print(f"even and odd sum are {even} , {odd}")
       
```

```python
## Print all factors are numbers

n = int(input("Enter the number :"))


for i in range(1,n+1):
    if n % i == 0:
        print(i)

```

```python
## Check if number is perfect or not

n = int(input("Enter the number :"))

perf = 0

for i in range(1 , n):
   if n % i == 0:
      perf = perf + i
print(perf)
         
```

```python
## Check Prime Number

n = int(input("Enter the number :"))

count = 0

for i in range(1,n+1):
    if n % i == 0:
        count = count + 1
if(count == 2):
    print("prime number")
else:
    print("not a prime number")

```
