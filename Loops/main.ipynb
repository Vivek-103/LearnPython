# Loops in Python


allows us to use a block of code multiple times without rewriting it .

## 2 Types of Loops

1. For Loop : When you know the number of iterations is Known
2. While Loop : You dont know number of iterations but you know a condition that determines when to stop.

## Terminiologies

1. RANGE FUNCTION : used to generate a sequence of numbers , commonly used in loops . Syntax : range ( start , stop , steps)
if not mentioned start value will be 0 and default steps will be 1 entering stop value is necessary

2. Loops for Numbers : For using loops with numbers you need the range function . EX : for i in range (1,6): ,print (i)

3. Loops for Strings : You can iterate though string in 2 ways:
    a. Using Index values
    b. Iterating directly over the string.
    EX : a = "nature" , for i in range(len(a)): print(a[i])-> iterated over string using index values
    EX : a = "nature" , for chat in a: print(char) -> directly iterate, but this method will give direct access to the characters instead of index values.

4. WHILE LOOP : The While loop repeats a block of code as long as condition is True . It is useful when the number of iterations is unknown before execution.
    While condition : -> code to execute

```python
## Check Prime Number

n = int(input("Enter the number :"))

count = 0

for i in range(1,n+1):
    if n % i == 0:
        count = count + 1
if(count == 2):
    print("prime number")
else:
    print("not a prime number")

```

```python
## Print Prime numbers within range

def print_primes_in_range(start, end):
    for num in range(start, end + 1):
        if num > 1:
            for i in range(2, int(num**0.5) + 1):
                if (num % i) == 0:
                    break
            else:
                print(num, end=" ")
print_primes_in_range(10, 50)
```

```python
## Count all digit , special symbols and letters

a = "sfdaogn1234#$#!"

char = 0
dig = 0
spchr = 0

for i in a:
    if i.isdigit():
        dig += 1
    elif i.isalpha():
        char += 1
    else:
        spchr += 1
print (f"Characters are: {char}, digits are : {dig} and special characters are : {spchr}")
```

```python
import random

num = random.randint(1,100)

tries = 0

while True:

    guess = int(input("Please guess your number b/w 1 and 100: "))

    tries += 1

    if num == guess:
        print(f"You are Right! You guessed the number in {tries} tries")
        break

    elif num < guess:
        print("Go a Little Lower!!")

    else:
        print("Go a Little Higher")
```
