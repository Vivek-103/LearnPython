# DICTIONARY

***SOME TERMINOLOGIES***

    1.Mutable : Mutable, you can change , add or remove key-value-pair.

    2.Duplicates : Keys must be unique , but you can have duplicates in values.

    3. Order : Follows insertion order

    4. Heterogenous :A dictionarary can store diffrent types of keys and values , like integers , strings, lists or even another dictionary

**SYNTAX AND WORKING**

    We have key and value pair to store value in dictionary and keys in dictionary act like index values that we use in List.

student = {"name":"John","age":20}
print(student["name]) -> #output--> John

    We can perform CRUD operations on values but not all keys cause the keys cannot be changed after creation.

**DICTIONARY TRAVERSING**

We can traverse both keys and values in dictionary , but default loop is set on keys and you can access the values because of keys.

```python
d  = {1:100,2:200,3:300,4:400}
## READ
print(d[1])
print(d[4])

d[1] = 21

print(d)## CAN UPDATE THE VALUES BUT CANNOT UPDATE KEY

## Creating 
d.update({5:43})

print(d)

d[6] = 21

print(d)

## Delete

del(d[1])
print(d)

```

```python
## Traversing

d  = {1:100,2:200,3:300,4:400}
for i in d:
    print(i) ## Iterates over Keys

for i in d:
    print(d[i]) ## Iterates over Values

for i in d.values():
    print(i) ## Another method to iterate over values



```

```python
## Dictionary Methods

d  = {1:100,2:200,3:300,4:400}

c = d.copy()## Shallow Copy ==> in changing c there will be no changes in d.

print(c)

d.clear() ## Clears the Dictionary

print(d)

c.get(3) # give value

c.items()

```

```python
## Merge 2 dictionaries

d1 = {1:100,2:200,3:300,4:400}

d2 ={1:10,5:10,6:30,7:40}

for i in d2:
    d1[i] = d2[i]

print(d1)

print(d1[1])
```

```python
## SUM all values in 2 Dictionary 

d1 = {1:100,2:200,3:300,4:400}
d2 = {1:10,5:10,6:30,7:40}

total = 0

for i in d1:
    total += d1[i] + d2.get(i,0)

print(total)
```

```python
## Sum 1 dictionary

d1 = {1:100,2:200,3:300,4:400}

sum = 0

for i in d1:
    sum = sum + d1[i]

print(sum)
```

```python
## Count frequency of each element in a list

a = [1,12,4,1,2,2,2,3,3,4,4,4,5,5,6,7,8]
d ={}
for i in a:
    if i in d.keys():
        d[i] += 1
    else :
        d[i] = 1

print(d)
```

```python
## Combine 2 dictionary for adding values for common keys 

d1 = {1:100,2:200,3:300,4:400}

d2 ={1:10,5:10,6:30,7:40}

for i in d2:
    if i in d1.keys():
        d1[i] += d2[i]
    else :
        d1[i] = d2 [i]

print(d1)
```
