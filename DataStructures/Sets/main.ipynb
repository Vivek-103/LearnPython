# SETS


***SOME TERMINOLOGIES***

    1.Mutable : Sets are mutable you can change
    
    2. Duplicates : Cannot have Duplicates in set
    
    3. Unordered : Sets are unordered and we can not access through index values

    4. Heterogenous : Set is semi-hertero , it can store some data types like strings , numbers but not everything
    

**HOW SET STORES VALUE IN STRINGS**

Each value in set is hashed using hash function (hash() in Python).
The Hash is used as an index to store the element in memory.
Only Mutable objects can be stores in a set
Since Hashing doesnot maintain an order , sets are unordered

**SET TREVERSING**

A set cannot be treversed using the index values cause it is unordered and has no index.

So many times it will give random values 

**SET METHODS**

    s = {1,2,3}

    s.add(4) -> Adds an element to the set
    s.remove(2) -> removes 2 
    s.dicard(5) -> removes 5 (error if not found)
    s.pop() -> removes random element
    s.clear() removes all element

**SPECIAL METHODS AND OPERATIONS B/W 2 SETS**
    
    A.union(B) ===> SHORTCUT =========> print (A | B)

    A.intersection(B) ===> SHORTCUT =========> print (A & B)

    A.diffrence(B) ===> SHORTCUT =========> print (A - B)

    A.symmetric_diffrence(B) ===> SHORTCUT =========> print (A ^ B)

```python
## SPECIAL OPERATIONS

a = {1,2,3,4}

b = {4,5,6,7}

union = a.union(b)
print("union are:" ,union)
intersection = a.intersection(b)
print("intersection are :" , intersection)
diffrence = a.difference(b)
print("diffrence is:", diffrence)
sym_diff = a.symmetric_difference(b)
print("symmetric diff :",sym_diff)

```
