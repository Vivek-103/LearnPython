# Tuples

***SOME TERMINOLOGIES***

    1. Immutable : cannot change value of tuple

    2. Duplicates : You can have duplicate values in tuple there are no restriction

    3. Ordered : Set are ordered and you can access them through index values
    
    4. Heterogenous : Set also have heterogenous nature and can have diffrent types of data structure in tuple

**Traversing in Tuple**
    Traversed in same manner as lists , but we can not change anything once its made
    Methods of tuple are :
        index = t.index(9) -> finds the index of first occurence of 9
        t.count(5) -> count occurences of 5

**NOTE : THERE ARE ONLY 2 METHODS IN TUPLE ONE FOR FINDING AND THE OTHER FOR COUNTING THE OCCURENCES OF AN ELEMENT**

```python
a = (1, 2 ,3,4)

print(type(a))
```

```python
a = (1,2,1,3,1,4,5,1,1)

index = a.index(1)
count = a.count(1)
print("index:",index)
print("counts of 1:", count)
```
