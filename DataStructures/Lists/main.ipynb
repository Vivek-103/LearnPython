# Lists

***SOME TERMINOLOGIES***

    1. Mutable : refers to whether an object's value can be changed after creation . List allows this

    2. Duplicates : Having save value occuring multiple times . List allows this

    3. Ordered : List maintains ordered data structure , maintains the sequence of elements as they were inserted . This means you can access elemetns using their position

    4. Heterogenous : List have heterogenous nature that means we can have multiple data types inside list. 

**LIST BASICS** 
    1. represented using square brackets (ex : fruits = ["apple","orange","banana"])
    2. It has indexing and slicing

**LIST TRAVERSING**
    1. It can be looped using index values and directly
    2. List has some methods to perform tasks

**List Methods** 
    numbers = [5,2,9,1,5,6]
   
    1. numbers.append(10) -> add 10 to the end
    2. numbers.insert(2,15) -> insert 15 at index 2
    3. numbers.extend (20,25,30) -> add multiple elements at the end
    4. numbers.remove (5) -> remove the first occuurance of 5
    5. numbers.pop(3) -> removes the element at index 3
    6. numbers.index(6) -> finds index of 6
    7. numbers.count(5) -> count occurences of 5
    8. numbers.sort() -> sorts in ascending order
    9. numbers, reverse() -> reverses the list order
    10. numbers.copy() -> create a copy of list
    11. numbers.clear() -> removes all elements from lists

```python
## Print positive and negative elements of a List

l = [-45,67,12,-69,32,-4]
print("Positive Elemetns are:")
for i in l:
    if i >= 0:
        print(i)
print("Negative elements are:")
for i in l:
    if i < 0:
        print(i)
```

```python
## Mean of List Elements

a = [12,13,43,55,7,34,21,87]



print("Mean of list elements is:")
sum = 0
for i in range(len(a)):
    sum += a[i]
sum = sum // len(a)
print(sum)
```
