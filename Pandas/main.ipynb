# Pandas: Merge vs Join vs Concat

## Overview

Pandas provides three primary methods to combine DataFrames:

1. `merge()` - SQL-style joins based on columns or indexes.
2. `join()` - Convenient method for joining DataFrames by index.
3. `concat()` - Stacks DataFrames along rows or columns.

---

# 1. merge()

## Purpose

Combine DataFrames using one or more common columns (keys).

### Syntax

```python
pd.merge(
    left,
    right,
    on=None,
    how='inner'
)
```

### Example

```python
import pandas as pd

df1 = pd.DataFrame({
    'id': [1, 2, 3],
    'name': ['A', 'B', 'C']
})

df2 = pd.DataFrame({
    'id': [1, 2, 4],
    'age': [20, 21, 22]
})

result = pd.merge(df1, df2, on='id')
print(result)
```

### Output

```text
   id name  age
0   1    A   20
1   2    B   21
```

## Join Types

### Inner Join

```python
pd.merge(df1, df2, on='id', how='inner')
```

Returns matching records only.

### Left Join

```python
pd.merge(df1, df2, on='id', how='left')
```

Returns all rows from the left DataFrame.

### Right Join

```python
pd.merge(df1, df2, on='id', how='right')
```

Returns all rows from the right DataFrame.

### Outer Join

```python
pd.merge(df1, df2, on='id', how='outer')
```

Returns all rows from both DataFrames.

---

# 2. join()

## Purpose

Join DataFrames primarily using indexes.

### Syntax

```python
df1.join(df2)
```

### Example

```python
df1 = pd.DataFrame({
    'name': ['A', 'B', 'C']
}, index=[1, 2, 3])

df2 = pd.DataFrame({
    'age': [20, 21, 22]
}, index=[1, 2, 4])

result = df1.join(df2)
print(result)
```

### Output

```text
  name   age
1    A  20.0
2    B  21.0
3    C   NaN
```

## Equivalent Merge

```python
df1.merge(
    df2,
    left_index=True,
    right_index=True
)
```

## When to Use

* DataFrames already have meaningful indexes.
* Simpler syntax than `merge()`.
* Quick index-based joins.

---

# 3. concat()

## Purpose

Combine DataFrames by stacking them vertically or horizontally.

---

## Row-wise Concatenation (Default)

### Example

```python
df1 = pd.DataFrame({'A': [1, 2]})
df2 = pd.DataFrame({'A': [3, 4]})

result = pd.concat([df1, df2])
print(result)
```

### Output

```text
   A
0  1
1  2
0  3
1  4
```

### Reset Index

```python
pd.concat([df1, df2], ignore_index=True)
```

Output:

```text
   A
0  1
1  2
2  3
3  4
```

---

## Column-wise Concatenation

### Example

```python
pd.concat([df1, df2], axis=1)
```

### Output

```text
   A  A
0  1  3
1  2  4
```

## When to Use

* Appending datasets.
* Combining monthly reports.
* Stacking rows.
* Combining columns side-by-side.

---

# Comparison Table

| Feature                               | merge() | join()  | concat() |
| ------------------------------------- | ------- | ------- | -------- |
| SQL-style joins                       | ✅       | Limited | ❌        |
| Based on columns                      | ✅       | ❌       | ❌        |
| Based on indexes                      | ✅       | ✅       | ✅        |
| Stack rows                            | ❌       | ❌       | ✅        |
| Stack columns                         | ❌       | ❌       | ✅        |
| Supports inner/left/right/outer joins | ✅       | ✅       | ❌        |
| Most flexible                         | ✅       | ❌       | ❌        |
| Easiest syntax                        | ❌       | ✅       | ✅        |

---

# Interview Cheat Sheet

### Use `merge()`

When matching records using a common key.

```python
pd.merge(customers, orders, on='customer_id')
```

---

### Use `join()`

When matching DataFrames using indexes.

```python
df1.join(df2)
```

---

### Use `concat()`

When stacking DataFrames.

```python
pd.concat([df1, df2])
```

---

# Memory Trick

```text
merge  -> Match records using columns
join   -> Match indexes
concat -> Stack DataFrames
```
