```python
### RENAME MONTH NAMES TO ENGLISH

df.head()
```

```python
df['month_new'] = df['month'].map({'Janeiro' : 'Jan',
                                   'Fevereiro' : 'Feb',
                                   'Março': 'Mar',
                                   'Abril' : 'April',
                                   'Maio' : 'May',
                                   'Junho' : 'June',
                                   'Julho' : 'July',
                                   'Agosto' : 'Aug',
                                   'Setembro' : 'Sep',
                                   'Outubro' : 'Oct',
                                   'Novembro' : 'Nov',
                                   'Dezembro' : 'Dec'})
```

```python
df.head()
```

```python
#### Total Number of Forest Fires Reported

df.shape

print("Total Number of Fires Reported : ", df.shape[0])
```

```python
### IN WHICH MONTH MAXIMUM NUMBER OF FOREST FIRES WERE REPORTED

data1 = df.groupby('month_new')['number'].sum().reset_index()
```

```python
data1
```

```python
plt.figure(figsize=(14,5))
sns.barplot(x='month_new',y='number',data=data1)
```

```python
### IN WHICH YEAR MAXIMUM NUMBER OF FOREST FIRES WERE REPORTED

data2 = df.groupby('year')['number'].sum().reset_index()
data2
```

```python
plt.figure(figsize=(16,5))
sns.barplot(data=data2,x='year',y='number')
```

```python
### IN WHICH STATE MAXIMUM NUMBER IF FOREST FIREST WERE REPORTED

data3 = df.groupby('state')['number'].sum().reset_index()
data3
```

```python
data3.loc[data3['number'].idxmax()]
```

```python
plt.figure(figsize=(15, 6))
sns.barplot(data=data3, x='state', y='number')

plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
### TOTAL NUMBER OF FOREST FIRES IN AMAZONAS STATE

df[df['state'] == 'Amazonas']['number'].sum()
```

```python
#### DISPLAY NUMBER OF FOREST FIREST IN AMAZONAS (Year - WISE)

data4 = df[df['state'] == 'Amazonas']
```

```python
data4
```

```python
data5 = data4.groupby('year')['number'].sum().reset_index()
```

```python
print("Forest Fires Year Wise in AMAZONAS STATE") 
```

```python
from matplotlib.pyplot import xticks

plt.figure(figsize=(14,8))
sns.barplot(data=data5,x='year',y='number')
plt.title('AMAZONIAN STATE FOREST REPORTED (YEAR-WISE)')
plt.xticks(rotation = 90)
plt.show()
```

```python
### DISPLAY NUMBER OF FOREST FIRES REPORTED IN AMAZONAS DAY WISE

data6 = df[df['state']=='Amazonas']
data6
```

```python
import calendar

day = data6.groupby(data6['date'].dt.day_of_week)['number'].sum().reset_index()

day['date'] = day['date'].map(lambda x: calendar.day_name[x])

day
```

```python
plt.figure(figsize=(10,6))
sns.barplot(data=day,x='date',y='number')
```

```python
#### FIND TOTAL NUMBER OF FIRES WERE REPORTED IN 2015 and Visualize Data Based on Each Month

data2015 = df[df['year']==2015]

data2015
```

```python
month_wise = data2015.groupby('month_new')['number'].sum().reset_index()
month_wise
```

```python
sns.barplot(data=month_wise,x='month_new',y='number')
```

```python
### AVERAGE NUMBER OF FIRES WERE REPORTED FROM HIGHEST TO LOWEST STATE WISE

data7 = df.groupby('state')['number'].mean().sort_values(ascending=False)
```

```python
data7 = data7.reset_index()
data7
```

```python
plt.figure(figsize=(16,10))
sns.barplot(data=data7,x='state',y='number')
plt.xticks(rotation = 90)
plt.show()
```
