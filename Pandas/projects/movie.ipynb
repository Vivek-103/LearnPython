```python
### Classify Movies Based on Rating [Excellent , Good and Average]

df.columns
```

```python
def rating(rating):
    if rating >= 8.0:
        return "Excellent"
    if rating >= 6.0:
        return "Good"
    else:
        return "Average"
```

```python
df['Rating_Cat'] = df['Rating'].apply(rating)
```

```python
df.head()
```

```python
df[['Title' , 'Rating_Cat']]
```

```python
### FIND UNIQUE VALUES FROM GENRE

df['Genre']
```

```python
list1 = []
for value in df['Genre']:
    list1.append(value.split(','))
```

```python
list1
```

```python
one_d = []
for item in list1:
    for item1 in item:
        one_d.append(item1)
```

```python
one_d
```

```python
uni_list = []
for item in one_d:
    if item not in uni_list:
        uni_list.append(item)
```
