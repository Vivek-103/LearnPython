```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,5,11)
x
y = x ** 2  
```

```python
y
```

```python
plt.plot(x,y)
plt.title("Test Title")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
```

```python
## SUBPLOTS IN MATPLOT LIB

plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,4)
plt.plot(x,y)
plt.subplot(2,2,2)
plt.plot(x**2,x**2)
plt.subplot(2,2,3)
plt.plot(y,x)

```

Matplotlib allows the aspect ratio , DPI and figure size to be specified when the Figure object is created. You can use the figsize and dpi keyword arguments .

figsize is a tuple of the within and height of the figure in inches dp is the dots-per-inch (pixel per inch)

```python
### OBJECTED ORIENTED WAY

fig = plt.figure()
axis1 = fig.add_axes([0.1,0.1,0.8,0.8])
axis1.plot(x,y)
axis2 = fig.add_axes([0.3,0.4,0.3,0.3])
axis2.plot(x,y)
```

```python
### TYPES OF PLOTS

plt.scatter(x,y)
```

```python
from random import sample
data = sample(range(1,1000),100)
plt.hist(data)
```

```python
plt.boxplot(x)
```

```python
### SAVING PLOTS

fig = plt.figure()
ax = fig.add_axes([0.1,0.1,0.8,0.8])
ax.plot(x,y)
plt.savefig("basic plot")
```
