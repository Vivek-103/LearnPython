```python
### LOADING THE LIBRARIES AND DATASET

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv('IPL.csv')

df.head()
```

```python
df.info()
```

```python
print(f" your rows are {df.shape[0]} and columns are {df.shape[1]}")
```

```python
df.isnull().sum()
```

```python
## WHICH TEAM WON THE MOST MATCHES

match_wins = df['match_winner'].value_counts()
match_wins
```

```python
from matplotlib.pyplot import viridis


sns.barplot(y=match_wins.index,x=match_wins.values,palette='viridis')
plt.title("Most Match Win by Team")
```

```python
## TOSS DECISION TRENDS

sns.countplot(x=df['toss_decision'],palette='magma')
plt.title("Toss Decision Trends")

```

```python
## TOSS WINNER VS MATCH WINNER

count = df[df['toss_winner'] == df['match_winner']]['match_id'].count()

percentage = (count * 100)/df.shape[0]
```

```python
print(percentage.round(2))
```

```python
### HOW DO TEAMS WINS (RUNS VS WICKET)

sns.countplot(x=df['won_by'])
plt.title("WON BY")
```

```python
### PLAYER OF THE MATCH AWARDS

count = df['player_of_the_match'].value_counts().head(10)
count
```

```python
sns.barplot(x=count.values , y = count.index, palette='magma')
plt.title("TOP 10 PLAYTERS WITH MAN OF THE MATCH HOLDERS")
```

```python
### 2 Top Scorers

high = df.groupby('top_scorer')['highscore'].sum().sort_values(ascending=False).head(2)
high
```

```python
high.plot(kind='barh')
```

```python
### 10 Best Bowling Figures

df['highest_wickets'] = df['best_bowling_figure'].apply(lambda x  :x.split('--')[0])
df['highest_wickets'] = df['highest_wickets'].astype(int)
top_bowlers = df.groupby('best_bowling')['highest_wickets'].sum().sort_values(ascending=False).head(10)
```

```python
top_bowlers
```

```python
top_bowlers.plot(kind='barh')
```

```python
venue_count = df['venue'].value_counts()
venue_count
```

```python
sns.barplot(y=venue_count.index,x=venue_count.values,palette='mako')
```

```python
### WHO WON THE HIGHEST MARGINS BY RUNS

df[df['won_by'] == 'Runs'].sort_values(by='margin',ascending=False).head(1)[['match_winner','margin']]
```

```python
### WHICH PLAYER HAS HIGHEST INDIVIDUAL SCORE

df[df['highscore'] == df['highscore'].max()][['top_scorer','highscore']]
```

```python
### WHICH BOWLER HAS THE BEST BOWLING FIGURES

df[df['highest_wickets']== df['highest_wickets'].max()][['best_bowling','best_bowling_figure']]
```
