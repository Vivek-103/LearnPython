# DISTRIBUTION PLOT

We use distribution plots to visualize the distribution of quantitative data.

These plots are:

* distplot/histplot
* jointplot
* pairplot
* rugplot

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df =sns.load_dataset('tips')
```

```python
df
```

```python
df['size'].unique()
```

```python
from statistics import kde

plt.subplot(1,2,1)
sns.histplot(df['total_bill'] , kde=True)
plt.subplot(1,2,2)
sns.histplot(df['tip'] , kde=True)
```

```python
## joint plot 

sns.jointplot(data=df ,x ='total_bill', y='tip',kind='scatter')

```

```python
sns.jointplot(data=df ,x ='total_bill', y='tip',kind='kde')
```

```python
sns.jointplot(data=df ,x ='total_bill', y='tip',kind='hex')
```

```python
sns.jointplot(data=df ,x ='total_bill', y='tip',kind='reg')
```

```python
sns.pairplot(df,hue='sex')
```

```python
sns.rugplot(df)
```
