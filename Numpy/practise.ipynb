# PRACTISING THE NUMPY 

```python
import numpy as np
```

```python
## VALID SUDOKU 
s =  np.array([
    [5,3,4,6,7,8,9,1,2],
    [6,7,2,1,9,5,3,4,8],
    [1,9,8,3,4,2,5,6,7],
    
    [8,5,9,7,6,1,4,2,3],
    [4,2,6,8,5,3,7,9,1],
    [7,1,3,9,2,4,8,5,6],
    
    [9,6,1,5,3,7,2,8,4],
    [2,8,7,4,1,9,6,3,5],
    [3,4,5,2,8,6,1,7,9],
])

```

```python
a = np.sum(s,axis=1)
```

```python
for i in a:
    if i != 45:
        print("Sudoku not Valid")
        break
print(" For ROWS ITS VALID SUDOKU")
        
```

```python
b = np.sum(s,axis = 0)
```

```python
for i in b:
    if i != 45:
        print("Sudoku not Valid")
        break
print(" For COLUMNS ITS VALID SUDOKU")
```

```python
for i in range (0,9,3):
    for j in range(0,9,3):
        print(s[i:i+3,j:j+3])
```

```python
for i in range (0,9,3):
    for j in range(0,9,3):
        n = s[i:i+3,j:j+3]
        print(n.sum())
```

#### DATA FRAMES PROBLEMS AND EXERCISES

```python
## Columns : Age, Math Marks , Science Marks
data = np.array([
    [18,95,78], #student 1
    [19,92,88], #student 2
    [17,76,95], #student 3
    [18,65,70], #student 4
    [20,90,85]  #student 5
])
```

```python
## GET THE SHAPE OF MATRIX

data.shape
```

```python
## FIND THE AVERAGE AGE OF STUDENTS

age = np.mean(data[:,0])
print(age)
```

```python
## MATHS MARKS OF ALL STUDENTS
data[:,1]   
```

```python
## HIGHEST SCIENCE MARKS

sc_max = np.max(data[:,2])
print(sc_max)
```

```python
## DETAILS OF THE STUDENT WHO HAVE SCORED MORE THAN 90 IN MATHS

data[data[:,1] > 90]
```

```python
## INCREASE MATHS MARKS OF ALL STUDENTS BY 5

data[:,1] += 5

data
```

```python
## FIND HOW MANY STUDENTS ARE YOUNGER THAN 19

len(data[data[:,0]<19])
```

```python
## CALCULATE THE AVERAGE MARKS IN EACH SUBJECT COLUMN VISE MEAN

np.mean(data[:,1:], axis=0)
```

```python
## GET DATA OF STUDENTS WHO SCORES AT LEAST 80 IN BOTH SUBJECTS

data[(data[:,1] >= 80) & (data[:,2] >= 80)]
```
