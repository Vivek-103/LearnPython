### DEEP AND SHALLOW COPY

```python
a = np.arange(1,21)
a
```

```python
slice = a[:5]
slice  = slice * 10
slice
```

```python
a # NO CHANGES IN A  when performing slicing operations , THIS IS EXAMPLE IF SHALLOW COPY
```

```python
b = a 
b[0] = 99
b
```

```python
a ## DEEP COPYING when copying to do shallow copy we use ,copy method
```

```python
b = a.copy()

b[0] = 23

b
```

```python
a ## NO CHANGES IN A EVEN COPYING
```
